# Understanding Midfielder Analysis

## Understanding Relative Strengths
Relative strengths are calculated by comparing each attribute to the midfielder's personal average. Here's how it works:

For each midfielder, we calculate their mean rating across all 4 attributes:
- Pass control
- Ball progression
- Defensive contribution
- Creative output

Then we subtract this mean from each attribute. For example:
If a midfielder has:
- Pass control: 8.5
- Ball progression: 9.0
- Defensive contribution: 7.5
- Creative output: 9.0
- Mean = 8.5

Their relative strengths would be:
- Pass control: 0.0 (average)
- Ball progression: +0.5 (strength)
- Defensive contribution: -1.0 (weakness)
- Creative output: +0.5 (strength)

This tells us what each midfielder is relatively better or worse at, regardless of their overall quality. A midfielder rated 7.0 and one rated 9.0 could have similar relative strengths despite different absolute skill levels.

## Why Use 4 Clusters?
The code uses 4 clusters because midfielders typically fall into four natural categories:

### Deep-Lying Playmaker:
- Stronger relative performance in pass control
- Good ball progression
- Above average defensive contribution
- Traditional regista/deep playmaker profile

### Box-to-Box:
- Well-balanced across all attributes
- Slightly higher defensive contribution
- Good all-round profile suitable for various systems
- Versatile midfielder type

### Advanced Playmaker:
- Excel in creative output and ball progression
- Usually lower defensive contribution
- Modern attacking midfielder profile
- Fits teams that dominate possession

### Defensive Midfielder:
-Strong defensive contribution
-Moderate pass control
-Lower creative output and ball progression

## Understanding the Visualizations

### Radar Chart:
- Shows the "signature" of each cluster
- Each axis represents one attribute
- Larger area = more pronounced characteristics
- Helps visualize the distinctive traits of each profile

### Correlation Heatmap:
- Shows relationships between attributes
- Positive values (red) = attributes that tend to go together
- Negative values (blue) = attributes that tend to oppose each other
- Helps understand which skills commonly develop together

### Box Plots:
- Show the distribution of each attribute within clusters
- The box shows the middle 50% of values
- The line in the box is the median
- Whiskers show the range
- Points beyond whiskers are outliers

## Practical Applications
This analysis can be used for:

1. Scouting:
   - Finding midfielders that match a team's playing style
   - Identifying specialists for specific tactical roles
   - Understanding how a midfielder might fit into different systems

2. Development:
   - Identifying areas where midfielders can improve
   - Understanding natural strengths to build upon
   - Planning specialized training programs

3. Tactical Planning:
   - Understanding how to best use a midfielder's strengths
   - Planning midfield partnerships
   - Adapting tactics to available personnel

4. Squad Building:
   - Ensuring balanced midfield options
   - Identifying gaps in midfield profiles
   - Planning transfer targets

## Why It Matters
Understanding relative strengths rather than just absolute ratings is crucial because:

1. It helps identify specialists vs generalists
2. Shows true playing style regardless of overall quality
3. Helps match midfielders to tactical systems
4. Identifies hidden strengths that raw numbers might miss
5. Enables better understanding of midfield partnerships
6. Helps predict adaptation to different playing styles

## Additional Insights
- Pass control and ball progression often correlate strongly, suggesting these skills develop together
- Defensive contribution often has a negative correlation with creative output, showing the specialist nature of these roles
- Box-to-Box midfielders show the most consistent distributions across attributes
- Advanced Playmakers show the highest variance in creative output
- Deep-Lying Playmakers typically show the most consistent pass control metrics

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
import os
os.environ["LOKY_MAX_CPU_COUNT"] = "4"  # Replace 4 with your actual number of cores

# Set style for visualizations
plt.style.use("seaborn-v0_8-whitegrid")

# Read the actual CSV file
data = pd.read_csv('consolidated_midfielder_analysis.csv')

# Define features for analysis
features = ['pass_control', 'ball_progression', 'defensive_contribution', 'creative_output']

# Calculate relative strengths
midfielder_means = data[features].mean(axis=1)
relative_strengths = pd.DataFrame()
for feature in features:
    relative_strengths[feature] = data[feature] - midfielder_means

# First, analyze optimal number of clusters
print("Analyzing optimal number of clusters...")
max_clusters = 10
inertias = []
silhouette_scores = []

# Scale the data for analysis
scaler = StandardScaler()
scaled_data = scaler.fit_transform(relative_strengths)
scaled_df = pd.DataFrame(scaled_data, columns=features)

# Calculate metrics for different numbers of clusters
K = range(2, max_clusters + 1)
for k in K:
    kmeans_test = KMeans(n_clusters=k, random_state=42)
    kmeans_test.fit(scaled_df)
    inertias.append(kmeans_test.inertia_)
    silhouette_scores.append(silhouette_score(scaled_df, kmeans_test.labels_))

# Plot validation metrics
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Elbow curve
ax1.plot(K, inertias, 'bx-')
ax1.set_xlabel('k (number of clusters)')
ax1.set_ylabel('Inertia')
ax1.set_title('Elbow Method for Optimal k')

# Silhouette scores
ax2.plot(K, silhouette_scores, 'rx-')
ax2.set_xlabel('k (number of clusters)')
ax2.set_ylabel('Silhouette Score')
ax2.set_title('Silhouette Analysis for Optimal k')

plt.tight_layout()
plt.show()

print("\nElbow Method Inertias:", inertias)
print("\nSilhouette Scores:", silhouette_scores)

# Proceed with clustering using n_clusters=4
n_clusters = 4
kmeans = KMeans(n_clusters=n_clusters, random_state=42)
labels = kmeans.fit_predict(relative_strengths)

# Add cluster labels to data
data['cluster'] = labels
profile_styles = ['Deep-Lying Playmaker', 'Box-to-Box', 'Advanced Playmaker', 'Defensive Midfielder']

# Print clusters and their members
print("\n=== Midfielder Clusters ===")
for i in range(n_clusters):
    print(f"\n{profile_styles[i]} Profile:")
    cluster_mids = data[data['cluster'] == i]
    for _, row in cluster_mids.iterrows():
        print(f"- {row['midfielder']} ({row['team']}) - {row['season_id']}")
    
    # Print relative strengths for this cluster
    print("\nCluster characteristics:")
    cluster_means = relative_strengths[data['cluster'] == i].mean()
    for feature in features:
        print(f"{feature}: {cluster_means[feature]:+.2f}")
    print("-" * 50)

# Create radar chart for cluster centers
angles = np.linspace(0, 2*np.pi, len(features), endpoint=False)
angles = np.concatenate((angles, [angles[0]]))

fig, ax = plt.subplots(figsize=(12, 12), subplot_kw=dict(projection='polar'))
for i in range(n_clusters):
    values = kmeans.cluster_centers_[i]
    values = np.concatenate((values, [values[0]]))
    ax.plot(angles, values, 'o-', linewidth=2, label=profile_styles[i])
    ax.fill(angles, values, alpha=0.25)

ax.set_xticks(angles[:-1])
ax.set_xticklabels([f.replace('_', ' ').title() for f in features])
ax.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))
plt.title("Midfielder Profile Types\nBased on Relative Strengths")
plt.show()

# Create correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(relative_strengths.corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix - Relative Strengths')
plt.show()

# Additional analysis: Distribution of attributes by cluster
fig, axes = plt.subplots(2, 2, figsize=(15, 15))
fig.suptitle('Distribution of Attributes by Cluster Profile')

for idx, feature in enumerate(features):
    ax = axes[idx//2, idx%2]
    sns.boxplot(data=data, x='cluster', y=feature, ax=ax)
    ax.set_xlabel('Cluster')
    ax.set_xticklabels(profile_styles)
    ax.set_title(feature.replace('_', ' ').title())

plt.tight_layout()
plt.show()

# Print summary statistics for each cluster
print("\n=== Cluster Summary Statistics ===")
for i in range(n_clusters):
    print(f"\n{profile_styles[i]} Profile Summary:")
    cluster_data = data[data['cluster'] == i]
    print(f"Number of midfielders: {len(cluster_data)}")
    print("\nAverage absolute values:")
    print(cluster_data[features].mean())
    print("\nStandard deviations:")
    print(cluster_data[features].std())
    print("-" * 50)

# 1. Standardize the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(relative_strengths)

# 2. Apply PCA for visualization
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

# 3. Create a DataFrame with PCA results
pca_df = pd.DataFrame(data=pca_result, columns=['PC1', 'PC2'])
pca_df['Cluster'] = data['cluster']  # Add cluster labels

# 4. Create the scatter plot
plt.figure(figsize=(10, 8))

# Define colors and markers for each cluster
colors = ['#dc2626', '#22c55e', '#2563eb', '#f59e0b']  # red, green, blue, orange
markers = ['D', 'o', 's', '^']  # diamond, circle, square, triangle
labels = ['Deep-Lying Playmaker', 'Box-to-Box', 'Advanced Playmaker', 'Defensive Midfielder']

# Plot each cluster
for i in range(n_clusters):
    mask = pca_df['Cluster'] == i
    plt.scatter(
        pca_df[mask]['PC1'], 
        pca_df[mask]['PC2'],
        c=colors[i],
        marker=markers[i],
        label=labels[i],
        alpha=0.7,
        s=100  # marker size
    )

# Customize the plot
plt.title('Midfielder Profile Clusters', fontsize=14, pad=20)
plt.xlabel('First Principal Component', fontsize=12)
plt.ylabel('Second Principal Component', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=10)

# Add a brief explanation of the components
explained_var = pca.explained_variance_ratio_
plt.figtext(0.99, 0.01, 
            f'Explained variance: PC1={explained_var[0]:.2%}, PC2={explained_var[1]:.2%}',
            ha='right', fontsize=8)

# Optional: Add cluster centers
kmeans_pca = pca.transform(kmeans.cluster_centers_)
plt.scatter(kmeans_pca[:, 0], kmeans_pca[:, 1], 
           c='black', marker='x', s=200, linewidths=3, 
           label='Cluster Centers')

plt.tight_layout()
plt.show()

# Print explained variance ratio
print("\nPrincipal Components Explained Variance Ratio:")
print(f"PC1: {explained_var[0]:.2%}")
print(f"PC2: {explained_var[1]:.2%}")

# Optional: Print cluster statistics
print("\nCluster Statistics:")
for i in range(n_clusters):
    print(f"\n{labels[i]}:")
    cluster_data = data[data['cluster'] == i]
    print(f"Number of players: {len(cluster_data)}")
    print("\nAverage attributes:")
    print(cluster_data[features].mean())

# Create DataFrame with cluster information
export_data = data[['midfielder', 'team', 'season_id'] + features].copy()
export_data['cluster_profile'] = data['cluster'].map({
    0: 'Deep-Lying Playmaker',
    1: 'Box-to-Box',
    2: 'Advanced Playmaker',
    3: 'Defensive Midfielder'
})

# Add relative strength calculations
for feature in features:
    export_data[f'relative_{feature}'] = relative_strengths[feature]

# Sort by cluster and midfielder name
export_data = export_data.sort_values(['cluster_profile', 'midfielder'])

# Save to CSV
output_filename = 'midfielder_clusters_analysis.csv'
export_data.to_csv(output_filename, index=False)
print(f"\nCluster analysis exported to {output_filename}")
print(f"Total midfielders analyzed: {len(export_data)}")
print("\nBreakdown by profile:")
print(export_data['cluster_profile'].value_counts().to_string())
```
