# Understanding Defender Analysis Methodology

## Overview
This documentation details a sophisticated analysis of football defenders using machine learning techniques, specifically clustering analysis, to identify distinct defender profiles based on their playing characteristics.

## Data Structure

### Input Features
The analysis uses five key defensive attributes:
- Box Defending
- Aerial Ability
- One-v-One Defending
- Defensive Pressure
- Ball Progression

### Data Processing
1. Initial data is loaded from a CSV file ('defender_analysis_all_leagues_20250130_024951.csv')
2. Relative strengths are calculated by subtracting the mean of all attributes from each individual attribute
3. Data is standardized using StandardScaler for consistent analysis

## Analysis Methodology

### Optimal Cluster Determination
The code implements two validation techniques to determine the optimal number of clusters:

1. **Elbow Method**
   - Calculates inertia (within-cluster sum of squares) for k=2 to k=10
   - Plots inertia against number of clusters
   - Helps identify the point where adding more clusters provides diminishing returns

2. **Silhouette Analysis**
   - Measures how similar objects are to their assigned cluster compared to other clusters
   - Higher scores indicate better-defined clusters
   - Plotted alongside elbow curve for comprehensive validation

### Cluster Analysis
The final analysis uses K-means clustering with 3 clusters, identifying distinct defender profiles:

1. Ball-Playing Defender
2. Defensive Stopper
3. Wing-Back

### Visualization Components

1. **Radar Chart**
   - Displays cluster centers for each profile
   - Shows relative strengths across all five attributes
   - Helps visualize the distinctive characteristics of each profile

2. **Attribute Distribution Boxplots**
   - Shows distribution of each attribute across clusters
   - Helps understand the variance within each profile
   - Identifies outliers and typical ranges

3. **PCA Visualization**
   - Reduces the 5-dimensional data to 2 dimensions
   - Creates scatter plot of defenders in reduced dimensional space
   - Includes cluster centers and explained variance ratios
   - Helps visualize cluster separation and distribution

## Statistical Analysis

For each cluster, the code calculates and reports:
1. Number of defenders in each profile
2. Average absolute values of attributes
3. Standard deviations
4. Relative strengths compared to the mean

## Output Generation

The analysis creates a comprehensive CSV export ('defender_clusters_analysis.csv') containing:
1. Player identification (name, team, season)
2. Raw attribute scores
3. Relative strength calculations
4. Assigned cluster profile

## Data Export Format
The final export includes:
- Defender name
- Team
- Season ID
- Original attribute scores
- Relative strength scores
- Cluster profile assignment

## Visualization Settings
- Uses seaborn-whitegrid style for consistent plotting
- Implements a color scheme for clusters:
  - Ball-Playing Defender: Red (#dc2626)
  - Defensive Stopper: Green (#22c55e)
  - Wing-Back: Blue (#2563eb)
- Uses distinct markers for different profiles in PCA plot:
  - Diamond: Ball-Playing Defender
  - Circle: Defensive Stopper
  - Square: Wing-Back

## Performance Considerations
- Environment variable "LOKY_MAX_CPU_COUNT" is set to manage computational resources
- Standardization is applied before clustering to ensure equal feature weighting
- PCA is used for dimensionality reduction and visualization while preserving data relationships

## Notes on Interpretation
- Relative strengths indicate how a defender's attributes compare to the average
- Cluster assignments represent the overall profile that best matches a defender's characteristics
- PCA explained variance ratio indicates how much of the original variation is captured in the 2D visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
import os
os.environ["LOKY_MAX_CPU_COUNT"] = "4"  # Replace 4 with your actual number of cores

# Set style for visualizations
plt.style.use("seaborn-v0_8-whitegrid")

# Read the actual CSV file
data = pd.read_csv('defender_analysis_all_leagues_20250130_024951.csv')

# Define features for analysis
features = ['box_defending', 'aerial_ability', 'one_v_one', 'defensive_pressure', 'ball_progression']

# Calculate relative strengths
defender_means = data[features].mean(axis=1)
relative_strengths = pd.DataFrame()
for feature in features:
    relative_strengths[feature] = data[feature] - defender_means

# First, analyze optimal number of clusters
print("Analyzing optimal number of clusters...")
max_clusters = 10
inertias = []
silhouette_scores = []

# Scale the data for analysis
scaler = StandardScaler()
scaled_data = scaler.fit_transform(relative_strengths)
scaled_df = pd.DataFrame(scaled_data, columns=features)

# Calculate metrics for different numbers of clusters
K = range(2, max_clusters + 1)
for k in K:
    kmeans_test = KMeans(n_clusters=k, random_state=42)
    kmeans_test.fit(scaled_df)
    inertias.append(kmeans_test.inertia_)
    silhouette_scores.append(silhouette_score(scaled_df, kmeans_test.labels_))

# Plot validation metrics
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Elbow curve
ax1.plot(K, inertias, 'bx-')
ax1.set_xlabel('k (number of clusters)')
ax1.set_ylabel('Inertia')
ax1.set_title('Elbow Method for Optimal k')

# Silhouette scores
ax2.plot(K, silhouette_scores, 'rx-')
ax2.set_xlabel('k (number of clusters)')
ax2.set_ylabel('Silhouette Score')
ax2.set_title('Silhouette Analysis for Optimal k')

plt.tight_layout()
plt.show()

print("\nElbow Method Inertias:", inertias)
print("\nSilhouette Scores:", silhouette_scores)

# Proceed with original clustering using n_clusters=3
n_clusters = 3
kmeans = KMeans(n_clusters=n_clusters, random_state=42)
labels = kmeans.fit_predict(relative_strengths)

# Add cluster labels to data
data['cluster'] = labels
profile_styles = ['Ball-Playing Defender', 'Defensive Stopper', 'Wing-Back']

# Print clusters and their members
print("\n=== Defender Clusters ===")
for i in range(n_clusters):
    print(f"\n{profile_styles[i]} Profile:")
    cluster_defs = data[data['cluster'] == i]
    for _, row in cluster_defs.iterrows():
        print(f"- {row['defender']} ({row['team']}) - {row['season_id']}")
    
    # Print relative strengths for this cluster
    print("\nCluster characteristics:")
    cluster_means = relative_strengths[data['cluster'] == i].mean()
    for feature in features:
        print(f"{feature}: {cluster_means[feature]:+.2f}")
    print("-" * 50)

# Create radar chart for cluster centers
angles = np.linspace(0, 2*np.pi, len(features), endpoint=False)
angles = np.concatenate((angles, [angles[0]]))

fig, ax = plt.subplots(figsize=(12, 12), subplot_kw=dict(projection='polar'))
for i in range(n_clusters):
    values = kmeans.cluster_centers_[i]
    values = np.concatenate((values, [values[0]]))
    ax.plot(angles, values, 'o-', linewidth=2, label=profile_styles[i])
    ax.fill(angles, values, alpha=0.25)

ax.set_xticks(angles[:-1])
ax.set_xticklabels([f.replace('_', ' ').title() for f in features])
ax.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))
plt.title("Defender Profile Types\nBased on Relative Strengths")
plt.show()


# Additional analysis: Distribution of attributes by cluster
fig, axes = plt.subplots(2, 3, figsize=(15, 15))
fig.suptitle('Distribution of Attributes by Cluster Profile')

for idx, feature in enumerate(features):
    ax = axes[idx//3, idx%3]
    sns.boxplot(data=data, x='cluster', y=feature, ax=ax)
    ax.set_xlabel('Cluster')
    ax.set_xticklabels(profile_styles)
    ax.set_title(feature.replace('_', ' ').title())

plt.tight_layout()
plt.show()

# Print summary statistics for each cluster
print("\n=== Cluster Summary Statistics ===")
for i in range(n_clusters):
    print(f"\n{profile_styles[i]} Profile Summary:")
    cluster_data = data[data['cluster'] == i]
    print(f"Number of defenders: {len(cluster_data)}")
    print("\nAverage absolute values:")
    print(cluster_data[features].mean())
    print("\nStandard deviations:")
    print(cluster_data[features].std())
    print("-" * 50)

# 1. Standardize the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(relative_strengths)

# 2. Apply PCA for visualization
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

# 3. Create a DataFrame with PCA results
pca_df = pd.DataFrame(data=pca_result, columns=['PC1', 'PC2'])
pca_df['Cluster'] = data['cluster']  # Add cluster labels

# 4. Create the scatter plot
plt.figure(figsize=(10, 8))

# Define colors and markers for each cluster
colors = ['#dc2626', '#22c55e', '#2563eb']  # red, green, blue
markers = ['D', 'o', 's']  # diamond, circle, square
labels = ['Ball-Playing Defender', 'Defensive Stopper', 'Wing-Back']

# Plot each cluster
for i in range(3):
    mask = pca_df['Cluster'] == i
    plt.scatter(
        pca_df[mask]['PC1'], 
        pca_df[mask]['PC2'],
        c=colors[i],
        marker=markers[i],
        label=labels[i],
        alpha=0.7,
        s=100  # marker size
    )

# Customize the plot
plt.title('Defender Profile Clusters', fontsize=14, pad=20)
plt.xlabel('First Principal Component', fontsize=12)
plt.ylabel('Second Principal Component', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=10)

# Add a brief explanation of the components
explained_var = pca.explained_variance_ratio_
plt.figtext(0.99, 0.01, 
            f'Explained variance: PC1={explained_var[0]:.2%}, PC2={explained_var[1]:.2%}',
            ha='right', fontsize=8)

# Optional: Add cluster centers
kmeans_pca = pca.transform(kmeans.cluster_centers_)
plt.scatter(kmeans_pca[:, 0], kmeans_pca[:, 1], 
           c='black', marker='x', s=200, linewidths=3, 
           label='Cluster Centers')

plt.tight_layout()
plt.show()

# Print explained variance ratio
print("\nPrincipal Components Explained Variance Ratio:")
print(f"PC1: {explained_var[0]:.2%}")
print(f"PC2: {explained_var[1]:.2%}")

# Optional: Print cluster statistics
print("\nCluster Statistics:")
for i in range(3):
    print(f"\n{labels[i]}:")
    cluster_data = data[data['cluster'] == i]
    print(f"Number of players: {len(cluster_data)}")
    print("\nAverage attributes:")
    print(cluster_data[features].mean())

# Create DataFrame with cluster information
export_data = data[['defender', 'team', 'season_id'] + features].copy()
export_data['cluster_profile'] = data['cluster'].map({
    0: 'Ball-Playing Defender',
    1: 'Defensive Stopper',
    2: 'Wing-Back'
})

# Add relative strength calculations
for feature in features:
    export_data[f'relative_{feature}'] = relative_strengths[feature]

# Sort by cluster and defender name
export_data = export_data.sort_values(['cluster_profile', 'defender'])

# Save to CSV
output_filename = 'defender_clusters_analysis.csv'
export_data.to_csv(output_filename, index=False)
print(f"\nCluster analysis exported to {output_filename}")
print(f"Total defenders analyzed: {len(export_data)}")
print("\nBreakdown by profile:")
print(export_data['cluster_profile'].value_counts().to_string())
```
