# Understanding Fullback Analysis Methodology

## Core Attributes Analyzed
The analysis focuses on five key attributes for fullbacks:
- Overlap/Inversion
- Final Third Entries
- Defensive Performance
- Ball Progression
- Chance Creation

## Understanding Relative Strengths
Relative strengths are calculated by comparing each attribute to the fullback's personal average. Here's how it works:

For each fullback, we calculate their mean rating across all 5 attributes, then subtract this mean from each attribute. For example:

If a fullback has:
- Overlap/Inversion: 8.5
- Final Third Entries: 7.5
- Defensive Performance: 8.0
- Ball Progression: 8.5
- Chance Creation: 7.5
Mean = 8.0

Their relative strengths would be:
- Overlap/Inversion: +0.5 (strength)
- Final Third Entries: -0.5 (weakness)
- Defensive Performance: 0.0 (average)
- Ball Progression: +0.5 (strength)
- Chance Creation: -0.5 (weakness)

This approach reveals what each fullback is relatively better or worse at, regardless of their overall quality. A fullback rated 6.5 and one rated 8.5 could have similar relative strengths despite different absolute skill levels.

## Clustering Analysis
The code implements clustering to identify distinct fullback profiles. Based on modern football trends, fullbacks typically fall into several categories:

### Defensive Specialists
- Stronger relative performance in defensive metrics
- More conservative in attacking phases
- Suited to teams that prioritize defensive stability

### Attacking Wing-Backs
- Excel in final third entries and chance creation
- More aggressive positioning
- Ideal for teams that use fullbacks as primary attacking outlets

### Progressive Playmakers
- Strong in ball progression and build-up play
- Balance between attack and defense
- Suited to possession-based systems

### Complete Fullbacks
- Well-rounded across all attributes
- No significant weaknesses
- Versatile enough to adapt to different tactical setups

## Understanding the Visualizations

### Radar Charts
- Display the "signature" of each cluster
- Each axis represents one attribute
- Larger area indicates more pronounced characteristics
- Helps visualize the distinctive traits of each profile
- Useful for comparing different fullback types

### Correlation Heatmap
- Shows relationships between attributes
- Positive correlations (red) indicate attributes that tend to develop together
- Negative correlations (blue) suggest trade-offs between different skills
- Helps understand which skills commonly complement each other

### Distribution Plots
- Show how attributes are distributed across the population
- Help identify what constitutes "normal" vs "exceptional" performance
- Useful for setting benchmarks and identifying outliers

### Box Plots
- Illustrate the distribution of attributes within clusters
- The box shows the middle 50% of values
- The line in the box represents the median
- Whiskers show the range of typical values
- Points beyond whiskers indicate unusual cases

## Principal Component Analysis (PCA)
The PCA analysis helps:
- Reduce the dimensionality of the data while preserving key information
- Identify the main patterns in fullback playing styles
- Understand which combinations of attributes explain most of the variation
- Visualize the relationships between different fullback profiles

## Implementation Notes
The analysis uses several sophisticated techniques:
- Standardization of metrics for fair comparison
- Clustering to identify natural groupings
- PCA for dimensionality reduction
- Visualization techniques for clear communication of findings

This comprehensive approach provides a robust framework for understanding and analyzing fullback performances and profiles in modern football.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
import os
os.environ["LOKY_MAX_CPU_COUNT"] = "4"  # Replace 4 with your actual number of cores

# Set style for visualizations
plt.style.use("seaborn-v0_8-whitegrid")

# Define features for analysis
features = ['overlap_inversion', 'final_third_entries', 'defensive_performance', 
           'ball_progression', 'chance_creation']

def calculate_relative_strengths(data, features):
    """Calculate relative strengths for each fullback."""
    fullback_means = data[features].mean(axis=1)
    relative_strengths = pd.DataFrame()
    for feature in features:
        relative_strengths[feature] = data[feature] - fullback_means
    return relative_strengths

def analyze_optimal_clusters(data, features, max_clusters=10):
    """Analyze optimal number of clusters using multiple methods."""
    # Prepare the data
    scaler = StandardScaler()
    relative_strengths = calculate_relative_strengths(data, features)
    
    # Scale the data
    relative_strengths_scaled = scaler.fit_transform(relative_strengths)
    relative_strengths_scaled = pd.DataFrame(relative_strengths_scaled, columns=features)
    
    # Lists to store metrics
    inertias = []
    silhouette_scores = []
    
    # Calculate metrics for different numbers of clusters
    K = range(2, max_clusters + 1)
    for k in K:
        kmeans = KMeans(n_clusters=k, random_state=42)
        kmeans.fit(relative_strengths_scaled)
        inertias.append(kmeans.inertia_)
        silhouette_scores.append(silhouette_score(relative_strengths_scaled, kmeans.labels_))
    
    # Plot validation metrics
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    
    # Elbow curve
    ax1.plot(K, inertias, 'bx-')
    ax1.set_xlabel('k (number of clusters)')
    ax1.set_ylabel('Inertia')
    ax1.set_title('Elbow Method for Optimal k')
    
    # Silhouette scores
    ax2.plot(K, silhouette_scores, 'rx-')
    ax2.set_xlabel('k (number of clusters)')
    ax2.set_ylabel('Silhouette Score')
    ax2.set_title('Silhouette Analysis for Optimal k')
    
    plt.tight_layout()
    plt.show()
    
    return {
        'k_values': list(K),
        'inertias': inertias,
        'silhouette_scores': silhouette_scores,
        'relative_strengths': relative_strengths_scaled,
        'scaler': scaler
    }

def analyze_fullback_profiles(data):
    """Analyze fullback profiles using clustering and relative strengths."""
    # First, analyze optimal number of clusters
    print("Analyzing optimal number of clusters...")
    cluster_analysis = analyze_optimal_clusters(data, features)
    
    # Calculate relative strengths
    relative_strengths = cluster_analysis['relative_strengths']
    
    # Perform clustering
    n_clusters = 4  # Four distinct fullback profiles
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    labels = kmeans.fit_predict(relative_strengths)
    
    # Add cluster labels to data
    data['cluster'] = labels
    profile_styles = ['Defensive Specialist', 'Attacking Wing-Back', 
                     'Progressive Playmaker', 'Complete Fullback']
    
    # Print clusters and their members
    print("\n=== Fullback Clusters ===")
    for i in range(n_clusters):
        print(f"\n{profile_styles[i]} Profile:")
        cluster_fbs = data[data['cluster'] == i]
        for _, row in cluster_fbs.iterrows():
            print(f"- {row['fullback']} ({row['team']}) - {row['season_id']}")
        
        # Print relative strengths for this cluster
        print("\nCluster characteristics:")
        cluster_means = relative_strengths[data['cluster'] == i].mean()
        for feature in features:
            print(f"{feature}: {cluster_means[feature]:+.2f}")
        print("-" * 50)
    
    # Create radar chart for cluster centers
    angles = np.linspace(0, 2*np.pi, len(features), endpoint=False)
    angles = np.concatenate((angles, [angles[0]]))
    
    fig, ax = plt.subplots(figsize=(12, 12), subplot_kw=dict(projection='polar'))
    for i in range(n_clusters):
        values = kmeans.cluster_centers_[i]
        values = np.concatenate((values, [values[0]]))
        ax.plot(angles, values, 'o-', linewidth=2, label=profile_styles[i])
        ax.fill(angles, values, alpha=0.25)
    
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels([f.replace('_', ' ').title() for f in features])
    ax.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))
    plt.title("Fullback Profile Types\nBased on Relative Strengths")
    plt.show()
    
    # Create correlation heatmap
    plt.figure(figsize=(10, 8))
    sns.heatmap(relative_strengths.corr(), annot=True, cmap='coolwarm', center=0)
    plt.title('Correlation Matrix - Relative Strengths')
    plt.show()
    
    # Distribution of attributes by cluster
    fig, axes = plt.subplots(3, 2, figsize=(15, 20))
    fig.suptitle('Distribution of Attributes by Cluster Profile')
    
    for idx, feature in enumerate(features):
        ax = axes[idx//2, idx%2]
        sns.boxplot(data=data, x='cluster', y=feature, ax=ax)
        ax.set_xlabel('Cluster')
        ax.set_xticklabels(profile_styles, rotation=45)
        ax.set_title(feature.replace('_', ' ').title())
    
    plt.tight_layout()
    plt.show()


    # Perform PCA analysis
    pca = PCA()
    pca_result = pca.fit_transform(StandardScaler().fit_transform(relative_strengths))

    # Plot explained variance ratio
    plt.figure(figsize=(10, 6))
    explained_var_ratio = pca.explained_variance_ratio_
    cum_explained_var = np.cumsum(explained_var_ratio)
    plt.plot(range(1, len(explained_var_ratio) + 1), cum_explained_var, 'bo-')
    plt.xlabel('Number of Components')
    plt.ylabel('Cumulative Explained Variance Ratio')
    plt.title('PCA Explained Variance Ratio')
    plt.grid(True)
    plt.show()

    # Print summary statistics for each cluster
    print("\n=== Cluster Summary Statistics ===")
    for i in range(n_clusters):
        print(f"\n{profile_styles[i]} Profile Summary:")
        cluster_data = data[data['cluster'] == i]
        print(f"Number of fullbacks: {len(cluster_data)}")
        print("\nAverage absolute values:")
        print(cluster_data[features].mean())
        print("\nStandard deviations:")
        print(cluster_data[features].std())
        print("-" * 50)

    
    # Return analysis results
    return relative_strengths, kmeans, labels






# Example usage:
if __name__ == "__main__":
    # Read the data
    data = pd.read_csv('combined_fullback_analysis.csv')
    
    # Perform analysis
    relative_strengths, kmeans, labels = analyze_fullback_profiles(data)
```
