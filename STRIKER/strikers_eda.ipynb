# Understanding Striker Analysis Methodology

## Overview
This documentation explains a comprehensive analysis system for classifying football strikers into distinct playing style profiles using machine learning techniques, specifically K-means clustering. The analysis considers seven key attributes and employs various statistical methods to identify and validate striker profiles.

## Key Components

### Data Requirements
The analysis expects a CSV file ('consolidated_striker_analysis.csv') containing the following features:
- player: Player name
- season_id: Season identifier
- minutes_played: Total minutes played
- matches_played: Number of matches participated
- avg_minutes_per_match: Average minutes per match
- Seven performance metrics:
  - finishing
  - early_shots
  - linkup_play
  - carrying
  - work_rate
  - back_to_goal
  - headed_duels

### Analysis Pipeline

#### 1. Data Preprocessing
- Calculation of relative strengths for each attribute
- Features are normalized using StandardScaler
- Missing values handling (if any)

#### 2. Optimal Cluster Analysis
The `analyze_optimal_clusters()` function performs multiple validation techniques:
- Elbow Method: Plots inertia vs. number of clusters
- Silhouette Analysis: Measures cluster cohesion and separation
- Supports analysis up to 10 clusters (configurable)
- Returns relative strengths dataframe for further analysis

#### 3. Clustering Implementation
- Utilizes K-means clustering with 4 clusters
- Assigns profile labels:
  1. Poacher
  2. Target Man
  3. Complete Forward
  4. Advanced Forward

#### 4. Visualization Suite

##### a. Radar Chart
- Displays cluster centers for each profile
- Shows relative strengths across all attributes
- Helps visualize distinctive characteristics of each profile

##### b. Correlation Heatmap
- Shows relationships between different attributes
- Uses coolwarm color mapping
- Includes numerical correlation coefficients

##### c. Distribution Boxplots
- Eight separate boxplots (2x4 grid)
- Shows attribute distribution by cluster
- Helps identify cluster-specific patterns

##### d. PCA Visualization
- Reduces dimensionality to 2 components
- Plots clusters in 2D space
- Marks cluster centers with red X
- Helps visualize cluster separation

### Statistical Analysis

#### 1. Cluster Profiling
For each cluster/profile:
- Lists all members with playing time
- Shows relative strengths in each attribute
- Provides cluster-specific characteristics

#### 2. Summary Statistics
For each profile type:
- Number of strikers in cluster
- Average values for all attributes
- Standard deviations
- Playing time analysis
  - Average minutes per match
  - Total matches in dataset

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
import os
os.environ["LOKY_MAX_CPU_COUNT"] = "4"  # Replace 4 with your actual number of cores

# Set style for visualizations
plt.style.use("seaborn-v0_8-whitegrid")

def analyze_optimal_clusters(data, features, max_clusters=10):
    """Analyze optimal number of clusters using multiple methods."""
    # Calculate relative strengths
    striker_means = data[features].mean(axis=1)
    relative_strengths = pd.DataFrame()
    for feature in features:
        relative_strengths[feature] = data[feature] - striker_means
    
    # Scale the data
    scaler = StandardScaler()
    relative_strengths_scaled = scaler.fit_transform(relative_strengths)
    relative_strengths_scaled = pd.DataFrame(relative_strengths_scaled, columns=features)
    
    # Lists to store metrics
    inertias = []
    silhouette_scores = []
    
    # Calculate metrics for different numbers of clusters
    K = range(2, max_clusters + 1)
    for k in K:
        kmeans = KMeans(n_clusters=k, random_state=42)
        kmeans.fit(relative_strengths_scaled)
        inertias.append(kmeans.inertia_)
        silhouette_scores.append(silhouette_score(relative_strengths_scaled, kmeans.labels_))
    
    # Plot validation metrics
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    
    # Elbow curve
    ax1.plot(K, inertias, 'bx-')
    ax1.set_xlabel('k (number of clusters)')
    ax1.set_ylabel('Inertia')
    ax1.set_title('Elbow Method for Optimal k')
    
    # Silhouette scores
    ax2.plot(K, silhouette_scores, 'rx-')
    ax2.set_xlabel('k (number of clusters)')
    ax2.set_ylabel('Silhouette Score')
    ax2.set_title('Silhouette Analysis for Optimal k')
    
    plt.tight_layout()
    plt.show()
    
    # Print the results
    print("\nClustering Validation Metrics:")
    print("\nElbow Method Results:")
    for k, inertia in zip(K, inertias):
        print(f"k={k}: Inertia={inertia:.2f}")
    
    print("\nSilhouette Analysis Results:")
    for k, score in zip(K, silhouette_scores):
        print(f"k={k}: Silhouette Score={score:.3f}")
    
    return relative_strengths

# Read the CSV file
data = pd.read_csv('consolidated_striker_analysis.csv')

# Define features for analysis
features = ['finishing', 'early_shots', 'linkup_play', 'carrying', 
           'work_rate', 'back_to_goal', 'headed_duels']

# First, analyze optimal number of clusters
print("Analyzing optimal number of clusters...")
relative_strengths = analyze_optimal_clusters(data, features)

# Continue with the existing clustering analysis using 4 clusters
n_clusters = 4
kmeans = KMeans(n_clusters=n_clusters, random_state=42)
labels = kmeans.fit_predict(relative_strengths)

# Add cluster labels to data
data['cluster'] = labels
profile_styles = ['Poacher', 'Target Man', 'Complete Forward', 'Advanced Forward']

# Print clusters and their members
print("\n=== Striker Clusters ===")
for i in range(n_clusters):
    print(f"\n{profile_styles[i]} Profile:")
    cluster_strikers = data[data['cluster'] == i]
    for _, row in cluster_strikers.iterrows():
        print(f"- {row['player']} - {row['season_id']} ({row['minutes_played']} mins)")
    
    # Print relative strengths for this cluster
    print("\nCluster characteristics:")
    cluster_means = relative_strengths[data['cluster'] == i].mean()
    for feature in features:
        print(f"{feature}: {cluster_means[feature]:+.2f}")
    print("-" * 50)

# Create radar chart for cluster centers
angles = np.linspace(0, 2*np.pi, len(features), endpoint=False)
angles = np.concatenate((angles, [angles[0]]))

fig, ax = plt.subplots(figsize=(12, 12), subplot_kw=dict(projection='polar'))
for i in range(n_clusters):
    values = kmeans.cluster_centers_[i]
    values = np.concatenate((values, [values[0]]))
    ax.plot(angles, values, 'o-', linewidth=2, label=profile_styles[i])
    ax.fill(angles, values, alpha=0.25)

ax.set_xticks(angles[:-1])
ax.set_xticklabels([f.replace('_', ' ').title() for f in features])
ax.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))
plt.title("Striker Profile Types\nBased on Relative Strengths")
plt.show()

# Create correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(relative_strengths.corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix - Relative Strengths')
plt.tight_layout()
plt.show()

# Distribution of attributes by cluster
fig, axes = plt.subplots(2, 4, figsize=(20, 12))
fig.suptitle('Distribution of Attributes by Cluster Profile')

for idx, feature in enumerate(features):
    ax = axes[idx//4, idx%4]
    sns.boxplot(data=data, x='cluster', y=feature, ax=ax)
    ax.set_xlabel('Cluster')
    ax.set_xticklabels(profile_styles, rotation=45)
    ax.set_title(feature.replace('_', ' ').title())

plt.tight_layout()
plt.show()

# Add cluster visualization using PCA
scaler = StandardScaler()
scaled_data = scaler.fit_transform(relative_strengths)
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

plt.figure(figsize=(12, 8))
scatter = plt.scatter(pca_result[:, 0], pca_result[:, 1], c=labels, cmap='viridis')
plt.colorbar(scatter, label='Cluster')
plt.xlabel('First Principal Component')
plt.ylabel('Second Principal Component')
plt.title('Striker Clusters Visualization')

# Add cluster centers
pca_centers = pca.transform(kmeans.cluster_centers_)
plt.scatter(pca_centers[:, 0], pca_centers[:, 1], c='red', marker='x', s=200, linewidth=3, label='Cluster Centers')
plt.legend()
plt.show()

# Print summary statistics for each cluster
print("\n=== Cluster Summary Statistics ===")
for i in range(n_clusters):
    print(f"\n{profile_styles[i]} Profile Summary:")
    cluster_data = data[data['cluster'] == i]
    print(f"Number of strikers: {len(cluster_data)}")
    print("\nAverage absolute values:")
    print(cluster_data[features].mean())
    print("\nStandard deviations:")
    print(cluster_data[features].std())
    print("-" * 50)

# Add playing time analysis
print("\n=== Playing Time Analysis ===")
for i in range(n_clusters):
    cluster_data = data[data['cluster'] == i]
    print(f"\n{profile_styles[i]} Profile:")
    print(f"Average minutes per match: {cluster_data['avg_minutes_per_match'].mean():.2f}")
    print(f"Total matches in dataset: {cluster_data['matches_played'].sum()}")
    print("-" * 50)
````
